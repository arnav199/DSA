```python
def intersection(self,headA,headB):
    l1=l2=0
    c1=headA
    c2=headB
    while c1:
        l1+=1
        c1=c1.next
    while c2:
        l2+=1
        c2=c2.next
    if l1>l2:
        diff=l2-l1
        while diff:
            headB=headB.next
            diff-=1
    else:
        diff=l1-l2
        while diff:
            headA=headA.next
            diff-=1
    while headA != headB:
        headA=headA.next
        headB=headB.next
    return headA
```
